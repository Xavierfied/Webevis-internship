```python
import cv2 as cv
import yaml
from fsspec.config import conf
from ultralytics import YOLO
```

```python
model = YOLO("yolo11n.pt")
```

```python
PATH = "img/image1-14.jpeg"
imgg = cv.imread(PATH)

original = imgg.copy()
```

```python
# cv.imshow("Original", imgg)
# cv.waitKey(0)
```

```python
result = model(imgg, conf=0.5)
```

```python


for box in result[0].boxes:

    height, width = imgg.shape[:2]


    #XYXY
    x1, y1, x2, y2 = box.xyxy[0].tolist()
    x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)

    # # XYXYn
    # x1n, y1n, x2n, y2n = box.xyxyn[0].tolist()
    # x1 = int(x1n * width)
    # y1 = int(y1n * height)
    # x2 = int(x2n * width)
    # y2 = int(y2n * height)


    # # XYWH
    # cx, cy, w, h = box.xywh[0].tolist()
    # cx, cy, w, h = int(cx), int(cy), int(w), int(h)
    # x1 = cx - w // 2
    # y1 = cy - h // 2
    # x2 = cx + w // 2
    # y2 = cy + h // 2


    cxn, cyn, wn, hn = box.xywhn[0].tolist()

    cx = int(cxn * width)
    cy = int(cyn * height)
    w  = int(wn  * width)
    h  = int(hn  * height)

    x1 = cx - w // 2
    y1 = cy - h // 2
    x2 = cx + w // 2
    y2 = cy + h // 2



    cv.rectangle(imgg, (x1, y1), (x2, y2), (200, 210, 190), 2)

    label = result[0].names[int(box.cls.item())]
    conf = box.conf.item()

    cv.rectangle(imgg, (x1, y1), (x2, y2), (200, 210, 190), 2)
    cv.putText(imgg, f"{label}: {conf:.2f}", (x1, y1), cv.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

```

```python
cv.imshow("Result", imgg)
cv.waitKey(0)
```
